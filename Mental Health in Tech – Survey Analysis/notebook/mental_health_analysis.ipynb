```python
#Cleaning Gender Column
import pandas as pd

#Loading data
df = pd.read_csv(r'C:\Users\dhatc\OneDrive\Documents\PROJECT\Mental Health in Tech – Survey Analysis\data\survey.csv')

#Clean gender
def clean_gender(g):
    g = str(g).strip().lower()
    if g in ['male', 'm', 'man']:
        return 'Male'
    elif g in ['female', 'f', 'woman']:
        return 'Female'
    else:
        return 'Other'

df['Gender'] = df['Gender'].apply(clean_gender)

#Check distribution
print(df['Gender'].value_counts())
```

```python
#Drop Missing & Filter Age
df = df.dropna(subset=['Age', 'Gender', 'self_employed', 'treatment'])
df = df[(df['Age'] > 15) & (df['Age'] < 80)]
df.reset_index(drop=True, inplace=True)
df.head()
```

```python
#Gender Distribution
plt.figure(figsize=(6,4))
sns.countplot(x='Gender', data=df, palette='Set2')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
#Mental Health Treatment by Gender
plt.figure(figsize=(7,5))
sns.countplot(x='Gender', hue='treatment', data=df, palette='Set1')
plt.title('Treatment by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(title='Treatment')
plt.show()
```

```python
#Age Distribution
plt.figure(figsize=(8,4))
plt.hist(df['Age'], bins=20, color='skyblue', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
#Quick Insight Summary
print("🔍 Most represented age group:", df['Age'].value_counts().idxmax())
print("🧠 % seeking treatment:", round(df['treatment'].value_counts(normalize=True)['Yes']*100, 2), "%")
print("👥 Gender distribution:\n", df['Gender'].value_counts())
```
